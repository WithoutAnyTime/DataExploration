```python
import numpy as np

# Metraż (m2)
X = np.array([25, 30, 45, 50, 65, 80, 95, 110]).reshape(-1, 1)
# Cena (tys. PLN)
y = np.array([180, 210, 310, 350, 430, 520, 610, 700])

X_mean = np.mean(X)
X_std = np.std(X)
y_mean = np.mean(y)
y_std = np.std(y)

X_scaled = (X - X_mean) / X_std
y_scaled = (y - y_mean) / y_std
```

```python
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
model_simple = keras.Sequential([
    keras.Input(shape=(1,)),
    layers.Dense(1)
])

model_simple.summary()

model_simple.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.001),
    loss='mse'
)
```

```python
from sklearn.model_selection import train_test_split

X_train_scaled, X_test_scaled, y_train_scaled, y_test_scaled = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

print("Scaled data split into training and testing sets:")
print(f"X_train_scaled shape: {X_train_scaled.shape}")
print(f"X_test_scaled shape: {X_test_scaled.shape}")
print(f"y_train_scaled shape: {y_train_scaled.shape}")
print(f"y_test_scaled shape: {y_test_scaled.shape}")
```

```python
history_simple = model_simple.fit(
    X_train_scaled, y_train_scaled,
    epochs=200,
    batch_size=32,
    validation_data=(X_test_scaled, y_test_scaled),
    verbose=0
)

print("Trenowanie zakończone!")

weights = model_simple.get_weights()
a_learned_scaled = weights[0][0][0]
b_learned_scaled = weights[1][0]

print(f"\nNauczone parametry (skalowane):")
print(f"  a (nachylenie) = {a_learned_scaled:.4f}") #waga
print(f"  b (wyraz wolny) = {b_learned_scaled:.4f}") #bias
```

```python
apartment_area = np.array([[70]])

apartment_area_scaled = (apartment_area - X_mean) / X_std

predicted_price_scaled = model_simple.predict(apartment_area_scaled)

predicted_price = (predicted_price_scaled * y_std) + y_mean

print(f"Przewidywana cena dla mieszkania o metrażu 70 m2: {predicted_price[0][0]:.2f} tys. PLN")
```
